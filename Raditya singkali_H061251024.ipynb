## 1. APA ITU (R-T-C-I-K-O-Q-V)??
1. R — >Role (Peran)
    Menentukan "topeng" atau identitas keahlian yang harus diambil oleh AI.

    Tujuan: Mengatur sudut pandang, tingkat keahlian teknis, dan gaya bahasa AI.

    Contoh: "Anda adalah seorang Ahli Geofisika Senior."


2. T — >Task (Tugas)
    Pernyataan singkat mengenai aksi atau instruksi utama yang wajib dilakukan AI.

    Tujuan: Menegaskan inti dari apa yang harus dikerjakan.

    Contoh: "Tulis skrip Python untuk memfilter sinyal seismik."


3. C — >Context (Konteks)
    Latar belakang masalah, situasi, atau konteks situasi mengapa tugas tersebut perlu dilakukan.

    Tujuan: Membantu AI memahami alasan di balik tugas dan audiens yang dituju.

    Contoh: "Sinyal seismik ini tercemar oleh derau ground roll frekuensi rendah di lapangan."


4. I — >Input (Data Masukan)
    Data mentah, informasi, parameter, atau teks pendukung yang disuapkan ke AI.

    Tujuan: Mencegah AI menebak-nebak variabel penting.

    Contoh: "Gunakan frekuensi sampling 1000 Hz dan rentang passband 10–60 Hz."


5. K — >Constraints (Batasan / Aturan Ketat)
    Aturan, larangan, atau batasan scope yang tidak boleh dilanggar oleh AI.

    Tujuan: Menjaga jawaban tetap fokus dan mencegah AI menghasilkan jawaban yang bertele-tele atau salah kaprah.

    Contoh: "Jangan gunakan library Obspy, gunakan hanya Numpy dan Scipy."


6. O — >Output (Format Luaran)
    Bentuk visual atau struktur dari jawaban akhir yang Anda inginkan.

    Tujuan: Agar jawaban langsung siap pakai tanpa perlu Anda reformat ulang.

    Contoh: "Tampilkan dalam bentuk kode Python lengkap beserta komentar di setiap baris."


7. Q — >Quality Criteria (Kriteria Kualitas)
    Indikator atau standar keberhasilan yang menentukan jawaban tersebut dianggap "bagus".

    Tujuan: Memaksa AI memberikan standar kualitas jawaban tinggi (high-quality output).

    Contoh: "Kode harus bebas bug, efisien secara memori, dan mudah dipahami."


8. V — >Verification (Langkah Verifikasi / Validasi)
    Instruksi kepada AI untuk mengecek dan memvalidasi kembali jawabannya sebelum disajikan.

    Tujuan: Meminimalkan risiko "halusinasi" atau kesalahan matematis/logika pada jawaban AI.

    Contoh: "Periksa kembali spektrum frekuensi hasil pemfilteran untuk memastikan derau tereduksi minimal 20 dB."

Mengapa Harus Menggunakan Framework Ini?
Menghindari Jawaban Generik: AI cenderung memberikan jawaban standar/umum jika diberi instruksi pendek. RTC IKO QV memberikan arahan yang presisi.
Efisiensi Waktu: Mengurangi kebutuhan revisi atau bertanya berulang kali (back-and-forth prompts).
Hasil yang Siap Pakai: Format luaran, standar kualitas, dan validasi sudah diatur di awal.



## 2. Penerapan 5 Topik Geofisika Menggunakan Framework RTC IKO QV

### 1). Prompt: Pengolahan Sinyal Seismik 1D
- **R (Role):** Anda adalah seorang Ahli Geofisika Komputasi dan pakar Digital Signal Processing (DSP).
- **T (Task):** Tulis kode Python lengkap untuk melakukan pemfilteran sinyal seismik sintetik menggunakan Bandpass Filter Butterworth.
- **C (Context):** Data seismik lapangan sering kali terkontaminasi oleh derau frekuensi rendah (ground roll) dan derau frekuensi tinggi (powerline noise), sehingga memerlukan pemfilteran frekuensi untuk meningkatkan signal-to-noise ratio (SNR).
- **I (Input):** 
  - Frekuensi sampling ($fs$): 1000 Hz
  - Rentang frekuensi sinyal utama: 10 Hz - 60 Hz
  - Durasi sinyal: 2 detik
  - Sinyal sintetik: Gabungan gelombang sinus 15 Hz, 40 Hz, derau 5 Hz, dan derau 120 Hz.
- **K (Constraints):** 
  - Gunakan hanya library standar data sains: `numpy`, `scipy.signal`, dan `matplotlib`.
  - Dilarang menggunakan pustaka khusus geofisika seperti `ObsPy`.
- **O (Output):** 
  1. Skrip Python utuh yang siap dijalankan di VSCode.
  2. Grafik perbandingan sinyal sebelum dan sesudah difilter beserta spektrum frekuensinya (FFT).
- **Q (Quality Criteria):** 
  - Kode harus berjalan tanpa *error* (*bug-free*).
  - Penjelasan langkah-langkah logika pemfilteran dicantumkan pada komentar kode (*inline comments*).
- **V (Verification):** Lakukan pengecekan spektrum frekuensi sinyal hasil filter untuk memastikan frekuensi < 10 Hz dan > 60 Hz tereduksi secara signifikan (minimal -20 dB).

### 2). Prompt: Interpretasi Zonal Well Log
- **R (Role):** Anda adalah seorang Senior Petrophysicist di perusahaan minyak dan gas.
- **T (Task):** Analisis data well log bawah permukaan untuk menentukan zona prospek hidrokarbon dan menghitung nilai kejenuhan air ($Sw$).
- **C (Context):** Tim geologi memerlukan evaluasi cepat pada Formasi "X" untuk menentukan interval kedalaman mana yang berpotensi sebagai reservoir batu pasir berpori berisi minyak/gas.
- **I (Input):**
  - Interval Kedalaman: 1500m - 1520m
  - Gamma Ray (GR): 35 API (Rata-rata)
  - Resistivitas Dalam ($Rt$): 45 Ohm.m
  - Porositas Neutron ($\phi_N$): 0.22
  - Porositas Densitas ($\phi_D$): 0.24
  - Salinitas Air Formasi ($Rw$): 0.05 Ohm.m pada suhu formasi.
- **K (Constraints):**
  - Gunakan Persamaan Archie standar untuk perhitungan $Sw$ ($a=1, m=2, n=2$).
  - Hindari penjelasan akademis yang terlalu bertele-tele; fokus pada angka keputusan teknik.
- **O (Output):**
  - Ringkasan Laporan Interpretasi dalam bentuk tabel (Kedalaman, Litologi, Porositas Efektif, $Sw$, dan Potensi Fluida).
  - Kesimpulan kualitatif (Reservoir Prospek / Non-Prospek).
- **Q (Quality Criteria):**
  - Hasil perhitungan $Sw$ mutlak akurat secara matematis sesuai rumus Archie.
  - Penentuan litologi konsisten dengan respons log GR rendah (< 50 API).
- **V (Verification):** Verifikasi apakah terdapat *crossover* pada log Densitas-Neutron ($\phi_D \ge \phi_N$) untuk memvalidasi indikasi gas/minyak sebelum menyimpulkan potensi hidrokarbon.

### 3).Prompt: Pemodelan Inversi 2D Anomali Gaya Berat
- **R (Role):** Anda adalah seorang Peneliti Geofisika Eksplorasi spesialis Metode Gaya Berat (Gravity).
- **T (Task):** Susun rancangan algoritma dan alur kerja (*workflow*) pemodelan inversi 2D untuk memetakan kedalaman batuan alas (*basement*).
- **C (Context):** Wilayah studi mengalami penurunan cekungan (*sedimentary basin*), dan data Anomali Bouguer Lengkap (ABL) perlu dipisahkan serta diinversi untuk mengetahui kontur topografi *basement*.
- **I (Input):**
  - Data Anomali Sisa (Residual Anomaly) sepanjang lintasan 10 km.
  - Kontras densitas batuan sedimen dan *basement*: $\Delta\rho = -0.35 \text{ g/cm}^3$.
  - Estimasi kedalaman rata-rata *basement*: 1.5 km.
- **K (Constraints):**
  - Penjelasan alur kerja maksimal 500 kata.
  - Gunakan skema inversi terorganisasi tanpa langsung memposting seluruh kode program.
- **O (Output):**
  - Diagram Alir (Flowchart) menggunakan sintaks **Mermaid**.
  - Penjelasan matematis singkat mengenai fungsi obyektif dan stabilisasi/regulerisasi (misal: *Occam/Tikhonov Regularization*).
- **Q (Quality Criteria):**
  - Sintaks Mermaid valid dan dapat ter-render langsung di VSCode Preview.
  - Penjelasan fisis mencakup masalah ambiguitas (*non-uniqueness*) pada metode potensi.
- **V (Verification):** Posisikan pengecekan pada konvergensi iterasi: pastikan nilai *Root Mean Square Error* (RMSE) antara kalkulasi dan observasi < 5%.

### 4).Prompt: Desain Parameter Akuisisi Geolistrik 2D
- **R (Role):** Anda adalah seorang Konsultan Geofisika Teknik dan Lingkungan.
- **T (Task):** Rancang konfigurasi dan parameter akuisisi geolistrik *Electrical Resistivity Tomography* (ERT) 2D untuk memetakan bidang sesar aktif dangkal.
- **C (Context):** Sebuah proyek infrastruktur jalan tol akan melintasi zona dugaan sesar aktif. Pemetaan diperlukan hingga kedalaman 40 meter untuk mengidentifikasi zona hancuran (*fracture zone*).
- **I (Input):**
  - Panjang lintasan yang tersedia: 240 meter.
  - Target Kedalaman Pemetaan: 40 - 50 meter.
  - Peralatan: Resistivitimeter multi-saluran dengan 48 elektroda.
- **K (Constraints):**
  - Jangan menyarankan konfigurasi yang tidak efisien secara waktu untuk jumlah 48 elektroda.
  - Pembahasan terbatas pada aspek teknis geofisika akuisisi.
- **O (Output):**
  - Tabel Rekomendasi Parameter (Jenis Konfigurasi, Spasi Elektroda, Kedalaman Penetrasi maksimum, Resolusi Lateral).
  - Justifikasi teknis pemilihan array (misal: *Wenner-Schlumberger* vs *Dipole-Dipole*).
- **Q (Quality Criteria):**
  - Rekomendasi spasi elektroda matematis realistis terhadap panjang kabel ($48 \text{ elektroda} \times \text{spasi} \le 240\text{m}$).
  - Analisis mampu menjelaskan *trade-off* antara penetrasi kedalaman dan resolusi lateral.
- **V (Verification):** Verifikasi rumus rasio penetrasi kedalaman ($zb$) sesuai dengan array yang dipilih untuk memastikan target kedalaman 40m tercapai.

### 5). Prompt: Analisis Prekursor Seismik Gunung Api
- **R (Role):** Anda adalah seorang Pengamat Gunung Api & Seismolog Gunung Api Senior dari lembaga PVMBG.
- **T (Task):** Buatkan panduan analisis sinyal gelombang seismik untuk membedakan jenis gempa gunung api sebagai indikator peningkatan status aktivitas vulkanologi.
- **C (Context):** Gunung Api "Y" mengalami peningkatan aktivitas. Tim di stasiun pemantau harus memilah sinyal seismik secara akurat untuk mendeteksi migrasi magma dari dapur magma ke permukaan.
- **I (Input):**
  - Karakteristik bentuk gelombang (*waveform*) dari 3 jenis gempa: Gempa Vulkanik-Dalam (VA), Gempa Vulkanik-Dangkal (VB), dan Gempa Tremor Menerus.
- **K (Constraints):**
  - Bahasa harus lugas, jelas, dan menggunakan standar operasional (SOP) laboratorium pemantauan.
  - Dilarang memasukkan variabel luar seperti respons evakuasi warga.
- **O (Output):**
  - Format Panduan Matriks Perbandingan (Jenis Gempa, Rentang FrekuensiDominan, Kedalaman Sumber, dan Implikasi Fisis Magmatik).
- **Q (Quality Criteria):**
  - Klasifikasi frekuensi gempa akurat secara teori (misal: VA 10-20 Hz, VB 1-5 Hz).
  - Bahasa teknis konsisten dengan terminologi vulkanologi internasional.
- **V (Verification):** Lakukan verifikasi ulang hubungan antara akumulasi gempa VB/Tremor dengan kenaikan magma mendekati permukaan (*shallow conduit*) untuk memastikan akurasi analisis prekursor.
