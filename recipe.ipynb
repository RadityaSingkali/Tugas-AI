* **C (Context):** Data seismik lapangan sering kali terkontaminasi oleh derau frekuensi rendah (*ground roll*) dan derau frekuensi tinggi (*powerline noise*), sehingga memerlukan pemfilteran frekuensi untuk meningkatkan *signal-to-noise ratio* (SNR)[cite: 1].
